# 🤖 Building an Interactive AI API Application

In this project, we will build a simple interactive application using **Gradio** that receives text from the user and sends it to an **AI API** to generate a summary.

Throughout this project, we will learn how to:

- Build a user interface with Gradio
- Receive user input
- Work with AI APIs
- Send `POST` requests using the `Requests` library
- Work with JSON responses
- Validate user inputs
- Handle basic API errors
- Connect a Gradio interface to a Python function

In this project, we use **OpenRouter** as an interface for accessing AI models.

### Step 1: import libraries

```python
import gradio as gr
import requests
```

### Step 2: Configuration

```python
# Define the openRouter AI endpoint
API_URL = 'https://openrouter.ai/api/v1/chat/completions'

# Select the free model router
MODEL_NAME = 'openrouter/free'

# Set the minimum characters
MIN_TEXT_LENGTH = 100

# Maximum time to wate for respunse
REQUEST_TIMEOUT = 60

# Display the installed library version

print("Gradio version:", gr.__version__)
print("Requests version:", requests.__version__)

print("\nProject configuration loaded successfully.")
```

### Step 3: API Request & Error Handling 

```python
def summarize_text(text, api_key):

    if not text or len(text.strip()) < MIN_TEXT_LENGTH:
        return 'Error:The text is very short '

    if not api_key.strip():
        return 'Error:Plese Enter API Key'

    headers = {"Authorization": f"Bearer {api_key.strip()}"}

    payload = {
        "model": MODEL_NAME,
        "messages": [
            {"role": "system", "content": "Summarize the text in its original language."},
            {"role": "user", "content": text}
        ]
    }

    try:
        response = requests.post(
            API_URL, headers=headers, json=payload,
            timeout=REQUEST_TIMEOUT
        )
        response.raise_for_status()
        return response.json()["choices"][0]["message"]["content"]

    except requests.RequestException:
        return 'API connection error'

    except (KeyError, IndexError, TypeError):
        return 'Invalid response'
```

### Step 4: Gradio interface

```python
# interface

with gr.Blocks(title="AI Text Summarizer") as demo:

    # Add the application title.
    gr.Markdown('Text summerize with Ai')

    # Create the input fields.
    text_input = gr.Textbox(
        label='Input text',
        placeholder='Please enter your text',
        lines=8
    )

    api_key_input = gr.Textbox(
        label="API Key",
        placeholder='Please enter your api key',
        type="password"
    )

    # Create the summarize button.
    summarize_button = gr.Button("Summerize it")

    # Create the output field.
    summary_output = gr.Textbox(
        label="Genereated summerize",
        lines=8
    )

    # Connect the button to the summarization function.
    summarize_button.click(
        fn=summarize_text,
        inputs=[text_input, api_key_input],
        outputs=summary_output
    )

# Launch the application.
demo.launch()
    
```
